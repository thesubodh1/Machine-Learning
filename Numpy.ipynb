```python
## Dot Product Example (nut butter Sales)
np.random.seed(0)
# number of jars sold
sales_amounts = np.random.randint(20,size=(5,3))
sales_amounts
```

```python
# create weekly saled daaframe
import pandas as pd
weekly_sales = pd.DataFrame(sales_amounts,index=["Mon","Tues","Wed","Thu","Friday"],columns=["Almond Butter","Peanut Butter","Cashew Butter"])
weekly_sales
```

```python
price = np.array([10,8,12])
price
```

```python
# create price DF
butter_prices = pd.DataFrame(price.reshape(1,3),index=["Price"],columns = ["Almond Butter","Peanut Butter","Cashew Butter"])
butter_prices
```

```python
total_sales = price.dot(sales_amounts)
```

```python
# shapes are not aligned lets traspose
total_sales = price.dot(sales_amounts.T)
total_sales
```

```python
weekly_sales["Total ($)"] = total_sales
```

```python
weekly_sales
```

```python
weekly_sales = weekly_sales.drop("Total ($)",axis=1)
```

```python
weekly_sales
```

```python
weekly_sales.shape,butter_prices.shape
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
```

```python
daily_sales
```

```python
weekly_sales["Total ($)"] = daily_sales.T
```

```python
weekly_sales
```
