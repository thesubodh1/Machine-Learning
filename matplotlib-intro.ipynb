# Matplot lib example workflow

```python
# 0. import matplotlib and get it ready for plotting
import matplotlib.pyplot as plt

# 1. prepare Data
x = [1,2,3,4]
y = [11,22,33,44]

# 2. setup plot
fig,ax = plt.subplots(figsize=(10,10)) # width and height (x,y)

# 3. plot.data
ax.plot(x,y)

# 4. CUstomize the plot
ax.set(title="Simple Plot",xlabel="x-axis",ylabel="y-axis")

#5. save & show
fig.savefig("test-plot.png")
```

```python
# my personal plot
import matplotlib.pyplot as plt

y = [100,500,200,900,150]
x = [5,10,15,20,25]
fig,ax = plt.subplots(figsize=(20,10))
ax.plot(x,y)
ax.set(title="My personal plot",xlabel="X-axis",ylabel="Y-axis")
fig.savefig("personal-plot.png")
```

## Making figures with numpy arrays

we want
* line plot
* scatter plot
* bar plot
* histogram
* subplot

```python
# create some data
x = np.linspace(0,10,100)
x[:10]
```

```python
# plot the data and create a line plot
fig,ax = plt.subplots(figsize=(10,10))
ax.plot(x,x**2);
```

```python
# plot the data and create Scatter using same data
fig,ax = plt.subplots()
ax.scatter(x,np.exp(x));
```

```python
# Another Scatter plot
fig,ax = plt.subplots()
ax.scatter(x,np.sin(x));
```

```python
# make a plot form a dict
nut_butter_prices = {"Almond butter":10,"Peanut Butter":8,"Cashew Butter":12}
fig,ax = plt.subplots()
ax.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax.set(title="Nut Butter Price Store",ylabel="Price ($)");
```
