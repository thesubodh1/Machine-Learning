# Matplot lib example workflow

```python
# 0. import matplotlib and get it ready for plotting
import matplotlib.pyplot as plt

# 1. prepare Data
x = [1,2,3,4]
y = [11,22,33,44]

# 2. setup plot
fig,ax = plt.subplots(figsize=(10,10)) # width and height (x,y)

# 3. plot.data
ax.plot(x,y)

# 4. CUstomize the plot
ax.set(title="Simple Plot",xlabel="x-axis",ylabel="y-axis")

#5. save & show
fig.savefig("test-plot.png")
```

```python
# my personal plot
import matplotlib.pyplot as plt

y = [100,500,200,900,150]
x = [5,10,15,20,25]
fig,ax = plt.subplots(figsize=(20,10))
ax.plot(x,y)
ax.set(title="My personal plot",xlabel="X-axis",ylabel="Y-axis")
fig.savefig("personal-plot.png")
```

## Making figures with numpy arrays

we want
* line plot
* scatter plot
* bar plot
* histogram
* subplot

```python
# create some data
x = np.linspace(0,10,100)
x[:10]
```

```python
# plot the data and create a line plot
fig,ax = plt.subplots(figsize=(10,10))
ax.plot(x,x**2);
```

```python
# plot the data and create Scatter using same data
fig,ax = plt.subplots()
ax.scatter(x,np.exp(x));
```

```python
# Another Scatter plot
fig,ax = plt.subplots()
ax.scatter(x,np.sin(x));
```

```python
# make a plot form a dict
nut_butter_prices = {"Almond butter":10,"Peanut Butter":8,"Cashew Butter":12}
fig,ax = plt.subplots()
ax.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax.set(title="Nut Butter Price Store",ylabel="Price ($)");
```

```python
# horizantal bar graph
fig,ax = plt.subplots()
ax.barh(list(nut_butter_prices.keys()),list(nut_butter_prices.values()));
```

```python
# Histogram
# make some data
x = np.random.randn(1000)
fig,ax = plt.subplots()
ax.hist(x);
```

# Two options for multiple plots(subplots)

```python
# Subplot option 1
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(ncols=2,nrows=2,figsize=(10,5))

# plot to each different axis
ax1.plot(x,x/2);
ax2.scatter(np.random.random(10),np.random.random(10));
ax3.bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax4.hist(np.random.randn(1000));
```

```python
# subplot option 2
fig,ax = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

# Plot to each dufferent index
ax[0,0].plot(x,x/2);
ax[0,1].scatter(np.random.random(10),np.random.random(10));
ax[1,0].bar(nut_butter_prices.keys(),nut_butter_prices.values());
ax[1,1].hist(np.random.randn(1000));
```
